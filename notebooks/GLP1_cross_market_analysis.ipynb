# GLP-1 Market Dynamics, Forecasting, and Companion Nutrition Analytics

This notebook is a thin analytical front end over the production modules in `src/glp1_market`. All commercial values are synthetic and are intended for portfolio demonstration only.


```python
from pathlib import Path
import sys

PROJECT_ROOT = Path.cwd().parent if Path.cwd().name == 'notebooks' else Path.cwd()
sys.path.insert(0, str(PROJECT_ROOT / 'src'))

import pandas as pd
from IPython.display import display
from glp1_market import ProjectConfig, run_pipeline
from glp1_market.visualization import (
    plot_dual_axis_market_trajectory,
    plot_market_share_evolution,
    plot_segment_forecast,
)
```

## Execute the reproducible end-to-end pipeline


```python
config = ProjectConfig(
    data_dir=PROJECT_ROOT / 'data',
    output_dir=PROJECT_ROOT / 'outputs',
)
results = run_pipeline(config)
print('Pipeline complete.')
```

## Inspect the integrated datasets


```python
display(results['pharma'].head(8))
display(results['supplements'].head(8))
display(results['unified'].head(10))
```

## Commercial KPI pack


```python
display(results['annual_market_share'].style.format({'TRx_Market_Share': '{:.1%}', 'TRx': '{:,.0f}'}))
display(results['brand_trx_cagr'].style.format({'CAGR': '{:.1%}', 'Start_Value': '{:,.0f}', 'End_Value': '{:,.0f}'}))
display(results['monthly_hhi'].tail(12))
```

## Validate the engineered cross-market adoption lag


```python
correlation_pivot = results['correlation'].pivot(
    index='TRx_Lead_Months',
    columns='Correlation_Type',
    values='Pearson_Correlation',
)
display(correlation_pivot.style.format('{:.3f}'))
```
